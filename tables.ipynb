```python
n = []
with pdfplumber.open("igcse.pdf") as pdf:
    for p in pdf.pages:
        for t in p.extract_tables():
            for r in t:
                n.append(r)

new_n = []
for entry in n:
    if entry == ['Syllabus/Component', 'Code', 'Duration', 'Date', 'Session'] or all(len(item)==0 for item in entry) or  any(item is None for item in entry):
        continue

    new_n.append(entry)

```

```python
from re import search as r_search
import datetime

def process_data(big_data: list) -> dict[dict]:
    """
    sample input: 
    [0] ['Accounting (Multiple Choice)', '0452/12', '1h 15m', 'Friday 06 November 2026', 'PM']
    [1] ['Accounting', '0452/22', '1h 45m', 'Thursday 08 October 2026', 'PM']
    [2] ['Additional Mathematics', '0606/12', '2h', 'Wednesday 30 September\n2026', 'PM']
    """
    return_dict = {}
    
    for item in big_data:
        code = item[1].split("/")

        hours = int(r_search(r'(\d+)h', item[2]).group(1)) if 'h' in item[2] else 0
        minutes = int(r_search(r'(\d+)m', item[2]).group(1)) if 'm' in item[2] else 0

        if code[0] not in return_dict:
            return_dict[code[0]] = [{"subj_name": item[0], 
                                "subj_component": code[1],
                                "duration": 60*hours + minutes, 
                                "date": datetime.datetime.strptime(item[3], "%A %d %B %Y").isoformat()}]
        else:
            return_dict[code[0]].append({"subj_name": item[0], 
                                "subj_component": code[1],
                                "duration": 60*hours + minutes, 
                                "date": datetime.datetime.strptime(item[3], "%A %d %B %Y").isoformat()})

    return return_dict

print(process_data(new_n))

    
```
